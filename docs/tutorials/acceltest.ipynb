# ACCEL TEST
This section is a hands-on tutorial on how to make a simple run.
We use the 51 Peg RV data available on [GitHub](https://github.com/ReddTea/astroemperor/tree/main/tests/datafiles/51Peg/RV).

```python
import numpy as np
import astroemperor as emp
import multiprocessing

np.random.seed(1234)

target_folder = 'acceltest'
my_cores = multiprocessing.cpu_count()
eng = 'reddemcee'
```

```python
sim = emp.Simulation()

sim.cores__ = my_cores
sim.acceleration = 2
sim.keplerian_parameterisation = 0
sim.sinusoid = True


sim.starmass = 1.109 # HD55693 http://exoplanets.astro.yale.edu/science/analysis/spocs_evol_95.php
sim.instrument_names_RV = ['SYNTH']

sim.jitter_prargs = [0, 10]
#sim.eccentricity_prargs = [0, 0.1]

if True:
    sim.add_condition(['Period True', 'limits', [1, 10]])
    sim.add_condition(['Amplitude True', 'limits', [10., 100.]])
    #sim.add_condition(['Phase True', 'limits', [0., 1.]])

    sim.add_condition(['Offset 1', 'limits', [-150, 0]])
    sim.add_condition(['Jitter 1', 'limits', [0.1, 15]])

    if sim.acceleration == 1:
        sim.add_condition(['Acceleration', 'limits', [0, 30]])
        sim.add_condition(['Acceleration', 'init_pos', [15, 17]])
    elif sim.acceleration == 2:
        #sim.add_condition(['Acceleration Order 1', 'limits', [-0.01, 0.01]])
        #sim.add_condition(['Acceleration Order 1', 'init_pos', [-0.001, 0.001]])

        #sim.add_condition(['Acceleration Order 1', 'limits', [0, 30]])
        #sim.add_condition(['Acceleration Order 1', 'init_pos', [15, 17]])

        sim.add_condition(['Acceleration Order 1', 'fixed', 0.0])

        sim.add_condition(['Acceleration Order 2', 'limits', [0, 30]])
        sim.add_condition(['Acceleration Order 2', 'init_pos', [15, 17]])


    sim.add_condition(['Period True', 'init_pos', [4, 4.2]])     # 4.1
    sim.add_condition(['Amplitude True', 'init_pos', [54, 56]])  # 55
    sim.add_condition(['Phase True', 'init_pos', [0.6, 0.8]])    # 0.7
    sim.add_condition(['Offset 1', 'init_pos', [11-60, 13-60]])
    sim.add_condition(['Jitter 1', 'init_pos', [4, 6]])
    #sim.add_condition(['Longitude 1', 'init_pos', [2.1, 3.]])

if True:
    sim.plot_posteriors['plot'] = True
    sim.plot_posteriors['modes'] = [0]
    sim.plot_posteriors['temps'] = [0]
    sim.plot_posteriors['format'] = 'png'

    sim.plot_rates['window'] = 10
    sim.plot_histograms['plot'] = False
    
    sim.plot_trace['plot'] = False

    sim.plot_betas['plot'] = True
    sim.plot_rates['plot'] = True
    sim.plot_rates['window'] = 10

    sim.plot_keplerian_model['uncertain'] = False
    sim.plot_all['format'] = 'png'
    sim.plot_all['paper_mode'] = True


sim.set_engine('reddemcee')
```

```python
sim.engine_config['setup'] = [12, 200, 2000, 1]  # ntemps, nwalkers, nsweeps, nsteps
# Manual ladder
sim.engine_config['betas'] = [1.0, 0.5413, 0.2946, 0.1554, 0.07623, 0.03745, 0.01788, 0.008435, 0.004067, 0.001508, 0.0003244, 0.0]

sim.engine_config['tsw_history'] = True
sim.engine_config['smd_history'] = True
sim.engine_config['adapt_tau'] = 200
sim.engine_config['adapt_nu'] = 1.5
sim.engine_config['adapt_mode'] = 'SAR'


sim.run_config['burnin'] = 0.5  # its in niter
```

```python
sim.load_data('acceltest')  # folder read from /datafiles/
sim.autorun(0, 0)
```
